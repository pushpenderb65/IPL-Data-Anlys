# MOST RUNS MATCH

```python


# 1. Top 10 rows nikalna
top_10_targets = df[['team1', 'team2', 'target_runs', 'winner']].nlargest(10, 'target_runs').copy()

# 2. Match ka naam banane ke liye team1 aur team2 ko jodna (jaise: RCB vs CSK)
top_10_targets['match'] = top_10_targets['team1'] + ' vs ' + top_10_targets['team2']

# 3. Chart ko ascending order mein sort karna taaki sabse bada target sabse upar aaye
top_10_targets = top_10_targets.sort_values(by='target_runs', ascending=True)

# 4. Plotting
plt.figure(figsize=(10, 6))
plt.barh(top_10_targets['match'], top_10_targets['target_runs'], color='skyblue')

plt.title('Top 10 Highest Target Runs in Matches', fontsize=14, fontweight='bold')
plt.xlabel('Target Runs', fontsize=12)
plt.ylabel('Match (Team 1 vs Team 2)', fontsize=12)

# Grid lines add karna taaki values read karne mein aasani ho
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

# WINNER

```python

# 1. Top 10 rows nikalna
top_10_targets = df[['team1', 'team2', 'target_runs', 'winner']].nlargest(10, 'target_runs').copy()

# 2. Match ka naam banana (jaise: RCB vs CSK)
top_10_targets['match'] = top_10_targets['team1'] + ' vs ' + top_10_targets['team2']

# 3. Chart ko ascending order mein sort karna
top_10_targets = top_10_targets.sort_values(by='target_runs', ascending=True)

# 4. Plotting
plt.figure(figsize=(11, 6))
bars = plt.barh(top_10_targets['match'], top_10_targets['target_runs'], color='skyblue')

plt.title('Top 10 Highest Target Runs & Their Winners', fontsize=14, fontweight='bold')
plt.xlabel('Target Runs', fontsize=12)
plt.ylabel('Match (Team 1 vs Team 2)', fontsize=12)

# 5. Har bar ke aage/sath mein Winner ka naam likhna
for bar, winner in zip(bars, top_10_targets['winner']):
    width = bar.get_width()
    plt.text(width + 2, bar.get_y() + bar.get_height()/2, 
             f' Winner: {winner}', 
             va='center', fontsize=10, fontweight='semibold', color='darkblue')

plt.xlim(0, top_10_targets['target_runs'].max() + 45) # Text ke liye thodi jagah chhodne ke liye
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python

# 1. Pehle venue aur team names ko clean kar lete hain (jo upar kiya tha)
venue_mapping = {
    'Arun Jaitley Stadium, Delhi': 'Arun Jaitley Stadium',
    'Feroz Shah Kotla': 'Arun Jaitley Stadium',
    'Brabourne Stadium, Mumbai': 'Brabourne Stadium',
    'Dr DY Patil Sports Academy, Mumbai': 'Dr DY Patil Sports Academy',
    'Dr. Y.S. Rajasekhara Reddy ACA-VDCA Cricket Stadium, Visakhapatnam': 'Dr. Y.S. Rajasekhara Reddy ACA-VDCA Cricket Stadium',
    'Eden Gardens, Kolkata': 'Eden Gardens',
    'Himachal Pradesh Cricket Association Stadium, Dharamsala': 'Himachal Pradesh Cricket Association Stadium',
    'M Chinnaswamy Stadium, Bengaluru': 'M Chinnaswamy Stadium',
    'M.Chinnaswamy Stadium': 'M Chinnaswamy Stadium',
    'MA Chidambaram Stadium, Chepauk': 'MA Chidambaram Stadium',
    'MA Chidambaram Stadium, Chepauk, Chennai': 'MA Chidambaram Stadium',
    'Maharaja Yadavindra Singh International Cricket Stadium, New Chandigarh': 'Maharaja Yadavindra Singh International Cricket Stadium, Mullanpur',
    'Maharashtra Cricket Association Stadium, Pune': 'Maharashtra Cricket Association Stadium',
    'Punjab Cricket Association IS Bindra Stadium, Mohali': 'Punjab Cricket Association IS Bindra Stadium',
    'Punjab Cricket Association IS Bindra Stadium, Mohali, Chandigarh': 'Punjab Cricket Association IS Bindra Stadium',
    'Punjab Cricket Association Stadium, Mohali': 'Punjab Cricket Association IS Bindra Stadium',
    'Rajiv Gandhi International Stadium, Uppal': 'Rajiv Gandhi International Stadium',
    'Rajiv Gandhi International Stadium, Uppal, Hyderabad': 'Rajiv Gandhi International Stadium',
    'Sawai Mansingh Stadium, Jaipur': 'Sawai Mansingh Stadium',
    'Wankhede Stadium, Mumbai': 'Wankhede Stadium'
}
df['venue'] = df['venue'].replace(venue_mapping)

name_mapping = {
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Rising Pune Supergiants': 'Rising Pune Supergiant'
}
df['winner'] = df['winner'].replace(name_mapping)

# 2. Har venue par total kitne matches khele gaye hain
total_matches_venue = df.groupby('venue').size().reset_index(name='total_matches')

# 3. Har venue par kis team ne kitne matches jeete hain
venue_winner_counts = df.groupby(['venue', 'winner']).size().reset_index(name='wins')

# 4. Har venue par sabse zyada jeetne wali team nikalna
top_team_per_venue = venue_winner_counts.loc[venue_winner_counts.groupby('venue')['wins'].idxmax()].copy()

# 5. Total matches aur top team ke data ko aapas mein merge karna
final_df = pd.merge(top_team_per_venue, total_matches_venue, on='venue')

# 6. Columns ko arrange karna: Venue, Total Matches, Best Team, Wins by that Team
final_df = final_df[['venue', 'total_matches', 'winner', 'wins']]
final_df = final_df.rename(columns={'winner': 'top_team', 'wins': 'top_team_wins'})

# 7. Clean output bina index ke print karna
print(final_df.to_string(index=False))
```
